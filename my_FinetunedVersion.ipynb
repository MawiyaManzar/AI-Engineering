```python
import torch
from unsloth import FastLanguageModel

MODEL_NAME = "HuggingFaceTB/SmolLM-135M"
MAX_SEQ_LENGTH=1024

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=MODEL_NAME,
    max_seq_length=MAX_SEQ_LENGTH,
    load_in_4bit=True
)

FastLanguageModel.for_inference(model)

def generate_text(prompt,max_new_tokens=80):
    inputs= tokenizer(prompt,return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            temperature=0.7,
            top_p=0.9,
        )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

prompt = "Write a short paragraph about quarterly revenue growth in a company report."
print(generate_text(prompt))
```

```python
# Test the base model on financial prompts
financial_prompts = [
    "The company reported total revenue of",
    "Risk factors include exposure to fluctuations in",
    "Diluted earnings per share for the fiscal year ended",
    "The Board of Directors declared a quarterly dividend of",
    "Our long-term debt obligations consist primarily of",
]

print("=" * 70)
print("BASE MODEL -- Financial Text Generation")
print("=" * 70)
for prompt in financial_prompts:
    completion = generate_text(prompt)
    print(f"\nPrompt: {prompt}")
    print(f"Output: {completion[:200]}")
    print("-" * 50)
    
```

```python
import math
def compute_perplexity(model, tokenizer, texts, max_length=512):
    """
    Perplexity = e^(average cross-entropy loss)
    Lower = model is less surprised by the text = better fit

    This is THE metric for CPT. If perplexity drops on domain text
    after training, the model learned the domain.
    """
    total_loss = 0
    total_tokens = 0

    model.eval()
    for text in texts:
        inputs = tokenizer(
            text,
            return_tensors="pt",
            truncation=True,
            max_length=max_length
        ).to(model.device)

        with torch.no_grad():
            outputs = model(**inputs, labels=inputs.input_ids)

        num_tokens = inputs.input_ids.shape[1]
        total_loss += outputs.loss.item() * num_tokens
        total_tokens += num_tokens

    avg_loss = total_loss / total_tokens
    return math.exp(avg_loss)
```

```python
sample_financial_texts = [
    "The company reported revenue growth of 12% year over year in the fourth quarter.",
    "Operating expenses increased due to higher research and development investments.",
    "Cash flow from operations remained stable despite macroeconomic uncertainty.",
    "The Board of Directors approved a quarterly dividend for common shareholders.",
    "Long-term debt obligations include notes payable and lease liabilities.",
]

baseline_ppl = compute_perplexity(model, tokenizer, sample_financial_texts)
print(f"Baseline perplexity: {baseline_ppl:.2f}")
print("Interpretation: lower perplexity means the model is less surprised by this kind of finance text.")
```

```python
import re


def clean_sec_filing(text):
    """Minimal cleanup for noisy filing text before chunking."""
    text = re.sub(r"\s+", " ", text)
    text = text.replace("\x00", " ")
    return text.strip()



def chunk_texts(text, chunk_size=80, overlap=20):
    """Split text into overlapping word chunks for CPT training."""
    words = text.split()
    if chunk_size <= 0:
        raise ValueError("chunk_size must be positive")
    if overlap < 0 or overlap >= chunk_size:
        raise ValueError("overlap must be >= 0 and smaller than chunk_size")

    step = chunk_size - overlap
    chunks = []
    for start in range(0, len(words), step):
        chunk_words = words[start:start + chunk_size]
        if not chunk_words:
            continue
        chunks.append(" ".join(chunk_words))
        if start + chunk_size >= len(words):
            break
    return chunks



raw_filing_text = """
    ITEM 2. MANAGEMENT'S DISCUSSION AND ANALYSIS

    Revenue increased 12% year over year due to stronger enterprise demand.
    Operating margin improved as cloud infrastructure costs declined.
    Cash flow from operations remained stable despite macroeconomic pressure.
"""

cleaned_text = clean_sec_filing(raw_filing_text)
chunks = chunk_texts(cleaned_text, chunk_size=10, overlap=2)

print("Cleaned text:")
print(cleaned_text)
print("\nChunk count:", len(chunks))
print("Expected chunk count: 4")
for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}: {chunk}")
```

```python
from datasets import Dataset
from trl import SFTTrainer
from transformers import TrainingArguments

# Reuse Step 3 output (`chunks`) as tiny CPT text corpus.
train_dataset = Dataset.from_dict({"text": chunks})

# Attach LoRA adapters for lightweight CPT training.
model = FastLanguageModel.get_peft_model(
    model,
    r=8,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0.0,
    bias="none",
)

training_args = TrainingArguments(
    output_dir="step4_lora_cpt",
    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    max_steps=20,            # tiny run to validate pipeline
    logging_steps=1,
    save_steps=20,
    fp16=not torch.cuda.is_bf16_supported(),
    bf16=torch.cuda.is_bf16_supported(),
    report_to="none",
)

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=train_dataset,
    dataset_text_field="text",
    max_seq_length=256,
    args=training_args,
)

print("Step 4 ready: starting tiny LoRA CPT run...")
train_result = trainer.train()
print("Step 4 complete. Train loss:", train_result.training_loss)
```
